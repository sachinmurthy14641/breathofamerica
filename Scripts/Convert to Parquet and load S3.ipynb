```python
import boto3
import pandas as pd
from io import BytesIO

# S3 setup
bucket_name = 'americasbreath123'
#file_key = 'LLCP2018ASC/LLCP2018.ASC'
#file_key = 'LLCP2019ASC/LLCP2019.ASC'  # or full path like 'folder/LLCP2018.ASC'
#file_key = 'LLCP2020ASC/LLCP2020.ASC'
#file_key = 'LLCP2021ASC/LLCP2021.ASC'
#file_key = 'LLCP2022ASC/LLCP2022.ASC'
file_key = 'LLCP2023ASC/LLCP2023.ASC'
#parquet_key = 'LLCP2018_subset.parquet'
#parquet_key = 'LLCP2019_subset.parquet'
#parquet_key = 'LLCP2020_subset.parquet'
#parquet_key = 'LLCP2021_subset.parquet'
#parquet_key = 'LLCP2022_subset.parquet'
parquet_key = 'LLCP2023_subset.parquet'

# Select only necessary columns and their widths
colspecs = [
    (0 ,2), (1402 ,1403), (87 ,88), (2064 ,2065), (2066 ,2067), (186 ,187), (203 ,205), (187 ,188), 
    (200 ,201), (1409 ,1415), (1455 ,1465), (1408 ,1409), (147 ,148), (140 ,141), (141 ,142), 
    (144 ,145), (224 ,225), (225 ,226), (226 ,227), (227 ,228), (112 ,113), (233 ,235), (107 ,109), 
    (110 ,111), (111 ,112), (109 ,110)
]

colnames = [
    "_STATE", "_URBSTAT", "SEXVAR", "_SEX", "_AGEG5YR", "EDUCA", "INCOME3", "RENTHOM1", "EMPLOY1",
    "_STSTR", "_WT2RAKE", "MSCODE", "HAVARTH4", "ASTHMA3", "ASTHNOW", "CHCCOPD3", "SMOKE100",
    "SMOKDAY2", "USENOW3", "ECIGNOW2", "EXERANY2", "AVEDRNK3", "PRIMINS1", "MEDCOST1", "CHECKUP1", "PERSDOC3"
]

# Read ASC file from S3
s3 = boto3.client('s3',verify=False)
obj = s3.get_object(Bucket=bucket_name, Key=file_key)
body = obj['Body'].read()

# Convert to DataFrame
df = pd.read_fwf(BytesIO(body), colspecs=colspecs, names=colnames)

# Write the DataFrame to a buffer in parquet format
buffer = BytesIO()
df.to_parquet(buffer, index=False)
buffer.seek(0)

# Upload parquet to S3
s3.put_object(Bucket=bucket_name, Key=parquet_key, Body=buffer)
print(f"Uploaded {parquet_key} to s3://{bucket_name}/")
```
