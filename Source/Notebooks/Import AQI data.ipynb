```
#Import the necessary libraries

import pandas as pd
import numpy as np
```

```
#Lets import all files into dataframes

aqi_2014 = pd.read_csv('breathofamerica/Data/epa_raw/annual_aqi_by_county_2014.csv')
aqi_2015 = pd.read_csv('breathofamerica/Data/epa_raw/annual_aqi_by_county_2015.csv')
aqi_2016 = pd.read_csv('breathofamerica/Data/epa_raw/annual_aqi_by_county_2016.csv')
aqi_2017 = pd.read_csv('breathofamerica/Data/epa_raw/annual_aqi_by_county_2017.csv')
aqi_2018 = pd.read_csv('breathofamerica/Data/epa_raw/annual_aqi_by_county_2018.csv')
aqi_2019 = pd.read_csv('breathofamerica/Data/epa_raw/annual_aqi_by_county_2019.csv')
aqi_2020 = pd.read_csv('breathofamerica/Data/epa_raw/annual_aqi_by_county_2020.csv')
aqi_2021 = pd.read_csv('breathofamerica/Data/epa_raw/annual_aqi_by_county_2021.csv')
aqi_2022 = pd.read_csv('breathofamerica/Data/epa_raw/annual_aqi_by_county_2022.csv')
aqi_2023 = pd.read_csv('breathofamerica/Data/epa_raw/annual_aqi_by_county_2023.csv')
aqi_2024 = pd.read_csv('breathofamerica/Data/epa_raw/annual_aqi_by_county_2024.csv')
```

> What are some good summary statistics to calculate for this

For the AQI dataset, here are some useful summary statistics to calculate:

1. **Central Tendency**:
   - Mean, Median, and Mode for numeric columns like `Days with AQI`, `Good Days`, `Moderate Days`, `Max AQI`, `Median AQI`, etc.

2. **Dispersion**:
   - Standard Deviation and Variance for columns like `Max AQI`, `90th Percentile AQI`, and `Median AQI`.
   - Range (Max - Min) for numeric columns.

3. **Distribution**:
   - Percentiles (e.g., 25th, 50th, 75th) for columns like `Max AQI` and `Median AQI`.

4. **Counts**:
   - Total counts for each category in categorical columns like `State` and `County`.
   - Count of missing or zero values in numeric columns.

5. **Aggregates**:
   - Sum of `Good Days`, `Moderate Days`, `Unhealthy Days`, etc., to understand the overall air quality trends.

6. **Ratios/Proportions**:
   - Proportion of `Good Days` to `Days with AQI`.
   - Proportion of `Unhealthy Days` to `Days with AQI`.

7. **Extremes**:
   - Minimum and Maximum values for columns like `Max AQI`, `Median AQI`, and `Days with AQI`.

8. **Yearly Trends**:
   - Grouped statistics by `Year` to observe trends over time.

9. **State/County Analysis**:
   - Grouped statistics by `State` or `County` to identify regional differences.

10. **Correlation**:
    - Correlation between variables like `Days Ozone`, `Days PM2.5`, and `Max AQI`.
